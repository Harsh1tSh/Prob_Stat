```python
import numpy as np
import scipy as sp
import scipy.special as sps

import matplotlib.pyplot as plt
```

```python
n = 100
p = 0.5
q = 1 - p

pxbinomial = []

for k in range(n+1):
  temp = sps.comb(n, k) * p**k * q**(n-k)
  pxbinomial.append(temp)

mu = n * p
sigma = np.sqrt(n * p * q)

X = np.arange(0, n , 0.01) #arranging in 0 to n but with jump of 0.01
pxnormal = (1/(sigma * np.sqrt(2*np.pi))) * np.exp(-((X - mu) ** 2)/ (2 * sigma ** 2))


k_values = np.arange(n + 1) # arranging in 0 to n
plt.plot(k_values, pxbinomial, 'o-', label='Binomial')
plt.plot(X, pxnormal, label = 'Normal')
plt.legend()
plt.xlim(0,n) # this line fixex the plot from o to n



```
